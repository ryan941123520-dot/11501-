## Part 1：完成函數

`*scores` 會將傳入的位置參數收集成 tuple；`**options` 會將關鍵字參數收集成 dictionary。函數會依設定調整分數，再計算平方、總分、平均與最高分。

```python
def analyze_scores(*scores, **options):
    """分析學生分數，回傳調整後分數、平方、總分、平均與最高分。"""
    multiplier = options.get('multiplier', 1)
    bonus = options.get('bonus', 0)

    adjusted = [score * multiplier + bonus for score in scores]
    squared = [score ** 2 for score in adjusted]
    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0
    max_score = max(adjusted) if adjusted else None

    return {
        'adjusted_scores': adjusted,
        'squared_scores': squared,
        'total': total,
        'average': average,
        'max_score': max_score
    }
```

## Part 2：呼叫函數

### 問題 1：使用 `*` 與 `**` 解包


```python
scores = [60, 70, 80]
settings = {
    'multiplier': 1.1,
    'bonus': 5
}

result = analyze_scores(*scores, **settings)
```

### 問題 2：印出 `result`


```python
print(result)
```

## Part 3：思考與分析

### 1. 函數內部 `scores` 的資料型態和值是什麼？

`scores` 的型態是 `tuple`。使用 `analyze_scores(*scores, **settings)` 呼叫時，串列 `[60, 70, 80]` 被解包成三個位置參數，所以函數內部的 `scores` 是 `(60, 70, 80)`。

### 2. 函數內部 `options` 的資料型態和值是什麼？

`options` 的型態是 `dict`。本例中其值為 `{'multiplier': 1.1, 'bonus': 5}`。

### 3. `score * multiplier` 與 `score ** 2` 有什麼不同？

`*` 是乘法，例如 `60 * 1.1` 得到 `66.0`；`**` 是次方，例如 `71.0 ** 2` 得到 `5041.0`。在函數中，先用乘法和加分值調整分數，再將調整後的分數平方。

### 4. 為什麼使用 `return`，而不只使用 `print()`？

`return` 會把計算結果交還給呼叫端，讓程式可以把結果存進變數、繼續運算或傳給其他函數。`print()` 只負責將內容顯示在畫面上，不會把該內容當作函數的回傳值。

### 5. 如果呼叫 `analyze_scores(60, 70, 80)`，沒有提供設定會怎樣？

程式仍可正常執行。`options.get('multiplier', 1)` 會使用預設倍率 `1`，`options.get('bonus', 0)` 會使用預設加分 `0`，因此分數不會改變，調整後分數為 `[60, 70, 80]`。

### 6. 如何加入最高分數 `max_score`？

已在 Part 1 的函數中加入 `max_score = max(adjusted) if adjusted else None`，並在回傳字典中新增 `'max_score': max_score`。若有分數，回傳調整後最高分；若沒有分數，則回傳 `None`。

### 7. 如果沒有傳入任何分數，會回傳什麼？

呼叫 `analyze_scores()` 時，`scores` 是空 tuple，`adjusted` 與 `squared` 都是空串列，`total` 是 `0`，`average` 是 `0`，`max_score` 是 `None`。程式以條件式避免對空串列計算平均或呼叫 `max()` 而發生錯誤。

## Part 4：挑戰題

原本的函數把位置參數收集到 `scores`，但 `scores_list=[60, 70, 80]` 是一個關鍵字參數，會被收進 `options`，而不是自動拆成 `scores`。因此需要在函數中讀取 `scores_list`，並將它與位置參數合併。

```python
def analyze_scores_flexible(*scores, **options):
    """同時支援位置分數與 scores_list 關鍵字參數。"""
    scores_list = options.pop('scores_list', [])
    all_scores = tuple(scores) + tuple(scores_list)

    multiplier = options.get('multiplier', 1)
    bonus = options.get('bonus', 0)

    adjusted = [score * multiplier + bonus for score in all_scores]
    squared = [score ** 2 for score in adjusted]
    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0
    max_score = max(adjusted) if adjusted else None

    return {
        'adjusted_scores': adjusted,
        'squared_scores': squared,
        'total': total,
        'average': average,
        'max_score': max_score
    }

challenge_result = analyze_scores_flexible(
    multiplier=1.1,
    bonus=5,
    scores_list=[60, 70, 80]
)
print(challenge_result)
```

### 挑戰題回答

1. `*scores` 是函數定義中的參數收集語法，會把額外的位置參數收集成 tuple；`scores_list` 則是一般的關鍵字參數名稱，這個例子傳入的是一個串列。

2. `scores_list=[60, 70, 80]` 會被 `**options` 收集到字典裡，原本的 `scores` 仍是空 tuple。原始函數只迭代 `scores`，所以不會自動處理 `options['scores_list']`。

3. 修改版先用 `options.pop('scores_list', [])` 取出分數串列，再把它與 `scores` 合併成 `all_scores`，後續就能沿用相同的調整與統計邏輯。

## 總結

`*` 在函數定義中可收集位置參數，在呼叫函數時可解包串列或 tuple；它也代表乘法。`**` 在函數定義中可收集關鍵字參數，在呼叫函數時可解包字典；它也代表次方。`return` 則將函數結果回傳給呼叫端。

例如：`analyze_scores(*[60, 70], **{'bonus': 5})` 使用 `*` 和 `**` 解包參數；而 `3 * 4` 是乘法，`3 ** 2` 是次方。
