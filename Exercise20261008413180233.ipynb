## 第一題：輸出 nn 乘法表（只能使用 `for`）

輸入正整數 `n`（1 ≤ n ≤ 9），輸出從 1 到 n 的乘法表。欄位使用 f-string 對齊。

````python
n = int(input('請輸入正整數 n（1～9）：'))

if 1 <= n <= 9:
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            print(f'{i:>2}×{j:>2}={i * j:>2}', end='   ')
        print()
else:
    print('輸入錯誤，請輸入 1～9 的正整數。')
````

## 第二題：輸出二進位 nn 乘法表（只能使用 `while`）

將 1 到 n 以三位二進位表示，乘積以六位二進位表示。迴圈只使用 `while`，並用 f-string 格式化與對齊。

````python
n = int(input('請輸入正整數 n（1～9）：'))

if 1 <= n <= 9:
    i = 1
    while i <= n:
        j = 1
        while j <= n:
            print(f'{i:03b}×{j:03b}={i * j:06b}', end='   ')
            j += 1
        print()
        i += 1
else:
    print('輸入錯誤，請輸入 1～9 的正整數。')
````
