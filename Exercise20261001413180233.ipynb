## 十二生肖與浮點數

## 十二生肖程式

使用者可輸入西元年或民國年，程式會換算後找出十二生肖。

```python
year = int(input('請輸入年份（西元或民國）：'))
ad_year = year + 1911 if year <= 115 else year
zodiac = ['猴', '雞', '狗', '豬', '鼠', '牛', '虎', '兔', '龍', '蛇', '馬', '羊']
animal = zodiac[(ad_year - 4) % 12]
print(f'{ad_year} 年是 {animal} 年。')
```

## 1. 0.1 的二進位表示法

0.1 不斷乘以 2：0.1×2=0.2 取 0；0.2×2=0.4 取 0；0.4×2=0.8 取 0；0.8×2=1.6 取 1；之後會回到 0.2，因此開始循環。

所以 **0.1(10) = 0.000110011001100110011...(2)**，是一個無限循環二進位小數。Python 的一般 float 無法精確儲存它，只能儲存最接近的可表示值。

## 2. 無限多位時 Python 如何解決？

Python 的 `float` 通常採用 IEEE 754 雙精度（double precision），共 64 bits。遇到無法有限表示的數值時，只能保留有限位數並捨入。IEEE 754 常見的捨入方式是 **round to nearest, ties to even（四捨六入五成雙）**。因此 0.1 儲存的是最接近它的二進位浮點數，而不是十進位 0.1 的完全精確表示。

## 3. Python 如何知道浮點數精度？

執行 `sys.float_info` 可以查看 Python 所使用浮點數的相關資訊。

```python
import sys
print(sys.float_info)
```

重點：`max` 是最大有限值；`min` 是最小正規化正數；`dig=15` 表示約 15 位十進位有效數字；`mant_dig=53` 是有效位元數；`epsilon≈2.22×10^-16` 是 1.0 與下一個可表示浮點數的差距；`radix=2` 表示二進位；`rounds=1` 通常代表 round to nearest。

## 4. 0.1 + 0.2 為什麼不是 0.3？怎麼解決？

```python
print(0.1 + 0.2)
```

原因是 0.1、0.2、0.3 在二進位浮點數中通常都不能精確表示，所以運算使用的是近似值。一般數值比較可使用 `math.isclose()`：

```python
import math
print(math.isclose(0.1 + 0.2, 0.3))
```

如果需要十進位精確計算，例如金融金額，可以使用 `Decimal`，並從字串建立：

```python
from decimal import Decimal
result = Decimal('0.1') + Decimal('0.2')
print(result)
print(result == Decimal('0.3'))
```

## 5. 對一般浮點數加法，在科學或金融上是好方法嗎？

不能一概而論。科學計算通常可以使用 IEEE 754 浮點數，但應考慮誤差，不能隨意用 `==` 判斷兩個浮點結果完全相等，可使用 `math.isclose()` 或依問題做誤差分析。金融計算若要求金額精確，單純使用 binary `float` 通常不是最佳選擇，較適合使用 `Decimal` 並明確指定精度與捨入規則，或使用整數表示最小貨幣單位。因此 `math.isclose()` 是解決浮點數「比較」問題的好方法，但不是金融精確計算的萬用解法。
