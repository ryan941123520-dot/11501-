## 第一部分：Python 如何表示數學物件？

| 數學概念 | Python 型別 | 範例 |
|---|---|---|
| 自然數 | `int` | `0`, `1`, `2` |
| 整數 | `int` | `-3`, `0`, `5` |
| 有理數 | `fractions.Fraction` | `Fraction(1, 3)` |
| 實數 | `float`（近似表示） | `3.14` |
| 複數 | `complex` | `2+3j` |
| 向量 | `list` / `tuple`，數值運算常用 NumPy array | `[1, 2, 3]` |
| 矩陣 | 巢狀 `list`，數值運算常用 NumPy array | `[[1,2],[3,4]]` |
| 集合 | `set` | `{1, 2, 3}` |
| 命題真假 | `bool` | `True`, `False` |

### 1. 哪些是 Python 的內建型別？

`int`、`float`、`complex`、`set`、`bool`、`list`、`tuple` 等都是 Python 內建型別。

### 2. 哪些需要額外模組才能較完整地表示？

有理數可以使用標準函式庫 `fractions.Fraction`。向量與矩陣雖然可以用內建的 `list` 表示，但若需要真正的數值線性代數運算，通常使用 NumPy 的 `ndarray`。

### 3. Python 的型別分類與數學中的分類是否完全一致？

不完全一致。數學中的集合是抽象概念，而 Python 型別是程式語言中的資料表示方式。例如 Python 沒有獨立的「自然數」型別，自然數和整數都可以用 `int` 表示；`float` 也只是有限精度的浮點近似，不等於所有實數。

## 第二部分：數學概念與 Python 型別

### 1. Python 是否為上述每一個數系都提供獨立型別？

不是。Python 主要提供 `int`、`float`、`complex` 等數值型別，但沒有分別建立自然數、整數、有理數、實數、複數五種完全對應的獨立型別。自然數與整數都可使用 `int`，有理數可使用 `Fraction`，實數通常以 `float` 做有限精度近似。

### 2. 為什麼自然數與整數都使用 `int` 表示？

因為自然數是整數的一部分，程式通常不需要為「非負整數」另外建立一個型別。使用同一個 `int` 可以簡化語言設計，也能讓正負整數使用一致的運算規則。

### 3. Python 為什麼提供 `complex` 型別？

複數在工程、科學計算、訊號處理與數學運算中很常見。內建 `complex` 可以直接表示實部和虛部，例如 `2+3j`，因此不必自行建立資料結構。

### 4. Python 為什麼沒有內建的有理數型別？

一般程式設計中，整數與浮點數已涵蓋大量需求；若需要精確分數，Python 標準函式庫提供 `fractions.Fraction`。因此不需要把有理數做成最基本的內建型別。

### 5. `1`、`1.0`、`1+0j` 的型別

```python
1          # int
1.0        # float
1+0j       # complex
```

三者數值看起來相近，但資料型別不同，因此可用的運算與儲存方式也不同。

## 第四部分：深入探討各種型別

### `int`

1. 是，屬於內建型別。
2. 主要物件結構是 `PyLongObject`。
3. 使用可變長度的整數數字陣列儲存數值，CPython 會以多個 digit 組合出大型整數。
4. 沒有一般意義上的固定 32-bit 或 64-bit 上限；實際上受可用記憶體與實作限制影響。
5. 數值變大時配置更多空間，因此可以表示非常大的整數。

### `float`

1. 是，屬於內建型別。
2. 主要物件結構是 `PyFloatObject`。
3. 通常使用 C 的 `double` 儲存 IEEE 754 雙精度浮點數。
4. 有固定的表示格式與範圍，通常為 64 bits。
5. 不會因數值變大而增加精度；超過可表示範圍可能變成 `inf`，過小則可能下溢。

### `complex`

1. 是，屬於內建型別。
2. 主要物件結構是 `PyComplexObject`。
3. 儲存實部與虛部兩個浮點數，通常各使用一個 C `double`。
4. 物件的數值表示大小固定，但其兩個浮點成分各受浮點數範圍與精度限制。
5. 不會像 `int` 一樣隨數值大小動態增加有效位數。

### `set`

1. 是，屬於內建型別。
2. 主要物件結構是 `PySetObject`。
3. 使用雜湊表儲存元素，透過 hash 值快速尋找。
4. 不是固定大小，集合元素增加時需要擴充內部表格。
5. CPython 會重新配置或擴大雜湊表以維持合理的查找效率。

### `bool`

1. 是，屬於內建型別。
2. `bool` 是 `int` 的子類別，布林物件使用 CPython 的整數物件機制；`True` 與 `False` 是固定的單例物件。
3. 只需要表示真或假。
4. 對使用者而言只有兩個布林值，不會像一般 `int` 一樣因數值變大而擴充。
5. 不需要動態擴充其真值表示。

### `list`

1. 是，屬於內建型別。
2. 主要物件結構是 `PyListObject`。
3. 內部以連續的 `PyObject*` 指標陣列保存元素。
4. 長度不是固定的。
5. 加入元素使容量不足時，CPython 會進行 over-allocation 並重新配置更大的儲存空間，以降低頻繁擴充的成本。

## 第五部分：Python 整數與浮點數

### 大型整數

執行：

```python
print(10**100)
print(10**1000)
```

兩個運算都可以正常執行，因為 Python 的 `int` 使用任意精度整數。

### 1. 為什麼 Python 能表示如此大的整數？

因為 Python `int` 不是固定使用 32-bit 或 64-bit 儲存，而是可以使用多個內部 digit 儲存一個大型整數。數字越大，通常需要的記憶體越多。

### 2. Python 的 `int` 是否具有固定的位元數限制？

沒有像一般固定寬度整數那樣的固定 32-bit 或 64-bit 上限。實際限制主要來自可用記憶體、執行環境以及 CPython 的實作限制。

### 3. Arbitrary Precision 是什麼？

任意精度表示數值所需的儲存空間可以依數值大小增加，而不是被固定的 32 或 64 bits 綁住。因此只要系統仍有足夠資源，就能處理非常大的整數。

### 4. CPython 使用什麼物件實作 `int`？

主要使用 `PyLongObject`。

### 5. Python 整數大小真正的限制是什麼？

主要是可用記憶體、運算時間，以及 CPython 對物件大小等實作上的限制。因此「unlimited precision」不是「無限大的數字可以免費儲存」，而是沒有固定的機器字長上限。

### 浮點數測試

```python
0.1 + 0.2
# 0.30000000000000004

0.1 + 0.2 == 0.3
# False

format(0.1, ".20f")
# '0.10000000000000000555'
```

### 1. 為什麼結果與數學上的答案不同？

因為十進位的 `0.1`、`0.2` 在二進位浮點格式中通常無法被有限位元精確表示，所以實際儲存的是非常接近它們的近似值。兩個近似值相加後會產生可觀察到的捨入誤差。

### 2. 浮點數是否能精確表示所有小數？

不能。有限位元的二進位浮點格式只能精確表示某些數，例如整數或某些二的負次方；很多十進位小數只能近似表示。

### 3. 什麼是浮點數的 precision？

Precision 是表示數值時能保留多少有效資訊或有效位數。對 IEEE 754 double 而言，通常約有 15 到 17 位十進位有效數字的精度。

### 4. 什麼是浮點數的 range？

Range 是浮點格式能表示的數值大小範圍，包含最小的非零數值到最大的有限數值。對一般 IEEE 754 binary64 而言，最大有限值約為 `1.7976931348623157e308`。

### 5. Precision 與 range 有何不同？

Range 關心「能表示多大或多小」，precision 關心「在這個範圍內能表示得多細」。一個格式可能有很大的 range，但仍然無法精確表示某些小數。

## 第六部分：AI 模型中的數值表示

### 1. AI 模型中的權重通常使用什麼型別儲存？

深度學習模型通常使用固定寬度的浮點或低精度數值格式，例如 FP32、FP16、BF16，以及在部分現代硬體與模型中使用 FP8。實際格式會依模型、訓練方法與硬體而不同。

### 2. FP64 是什麼？

FP64 是 64-bit 浮點格式，也就是 IEEE 754 binary64。它具有較高的精度與很大的數值範圍，但每個數值需要 8 bytes，因此記憶體與計算成本較高。

### 3. FP32 是什麼？

FP32 是 32-bit 浮點格式，也就是 IEEE 754 binary32。每個數值需要 4 bytes，是科學計算與深度學習中很常見的格式。

### 4. FP16 是什麼？

FP16 是 16-bit 浮點格式，每個數值只需要 2 bytes。相較 FP32 可節省記憶體並提高部分硬體上的吞吐量，但有效精度與可表示範圍較小。

### 5. BF16 是什麼？

BF16 是 Brain Floating Point 16-bit 格式，也是 16 bits。它保留與 FP32 相同的 8-bit exponent，因此數值範圍接近 FP32，但只有 7-bit fraction，因此精度比 FP32 低。它特別適合深度學習中的矩陣運算。

### 6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？

因為神經網路權重通常需要連續的實數近似值，而不是任意精度整數。Python `int` 是物件型別，具有額外的物件與記憶體管理成本；它也不適合 GPU 大量平行的固定寬度矩陣運算。固定寬度的 FP16、BF16、FP32 等格式能讓硬體更有效率地進行大量運算並降低記憶體頻寬需求。

## 第七部分：比較不同數值表示方式

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|---|---:|---|---|
| `int` | 可變，非固定位元數 | 整數本身通常可精確表示；限制主要來自資源 | 一般程式的整數計算 |
| FP64 | 64 | 有浮點捨入誤差 | 高精度科學計算、數值分析 |
| FP32 | 32 | 有 | 科學計算、深度學習、一般 GPU 運算 |
| FP16 | 16 | 有，且精度與範圍較小 | 深度學習訓練、推論與 GPU 加速 |
| BF16 | 16 | 有 | 深度學習訓練與推論 |

### 1. 哪種表示方式最精確？

在這張表列出的浮點格式中，FP64 通常具有最高精度。若限定為整數且只表示整數值，任意精度 `int` 可以精確表示整數，但它與浮點數的「有效位數」概念不同，因此不能直接用同一標準比較。

### 2. 哪種表示方式最節省記憶體？

FP16 與 BF16 都是 16 bits，也就是每個數值 2 bytes，因此在這幾種固定寬度格式中最省記憶體。

### 3. 哪種表示方式最常見於 AI 訓練？

在題目列出的格式中，FP16 與 BF16 都很常見。現代 AI 訓練尤其常見 BF16，因為它保留較大的 exponent 範圍；實際選擇仍取決於 GPU、模型與訓練框架。

### 4. 哪種表示方式最常見於 AI 推論？

FP16 與 BF16 都常用；實際部署也可能使用 FP8 或量化的 INT8、INT4 等格式。若只比較本題列出的五種格式，FP16/BF16 都是常見選項。

### 思考題一

#### 1. Unlimited Precision 是什麼意思？

它表示 Python `int` 沒有固定的 32-bit 或 64-bit 精度上限。當整數變大時，可以配置更多儲存空間來保存它。

#### 2. 是否代表可以儲存無限大的整數？

不是。它表示沒有固定的機器字長上限，不代表實際電腦有無限記憶體。當數字大到需要的記憶體超過系統可提供的資源時，程式仍然無法儲存或計算。

#### 3. CPython 如何實現？

CPython 使用 `PyLongObject` 表示整數，透過可變長度的內部數字陣列保存大型整數。數值需要更多位元時，就使用更多 digit，因此能夠支援任意精度的概念。

### 思考題二

#### 1. 為什麼 `int` 幾乎沒有固定上限？

因為 `int` 採用可變長度的任意精度表示法，不受單一固定大小的 CPU 整數型別限制。

#### 2. 為什麼 `float` 存在範圍限制？

因為浮點數使用固定數量的 bits 分配給 sign、exponent 與 fraction。例如 binary64 總共只有 64 bits，因此 exponent 與 fraction 的大小都是固定的。

#### 3. 為什麼 `float` 可能產生誤差？

有限位元不可能精確表示所有實數，而十進位小數轉換成二進位浮點數時常需要捨入，所以運算過程會累積或暴露捨入誤差。

### 思考題三

#### 1. 兩種設計理念的差異

Python `int` 以「精確表示整數」為主要目標，因此願意使用更多記憶體和運算時間。FP16/BF16 則以「固定大小、快速且適合大量平行運算」為主要目標，因此接受有限精度。

#### 2. 各自的優點與缺點

**任意精度 `int`：**
- 優點：整數計算精確、沒有固定 32/64-bit 上限。
- 缺點：數值越大需要越多記憶體，運算通常也越慢。

**FP16/BF16：**
- 優點：每個數值只需 16 bits，能降低記憶體與記憶體頻寬需求，且適合 GPU/Tensor Core 等硬體。
- 缺點：精度與可表示資訊有限，可能出現捨入、overflow 或 underflow，因此訓練時需要適當的數值穩定技巧。

#### 3. 為什麼大型語言模型通常選擇較低精度？

因為模型可能有數十億甚至更多參數。若每個參數從 FP32 改成 16-bit 格式，單純權重儲存量大約可減半。更小的資料也能降低 GPU 記憶體與記憶體頻寬壓力，並讓支援低精度矩陣運算的硬體達到更高吞吐量。

#### 4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？

每個參數需要 64 bits，權重記憶體相較 FP32 約增加一倍、相較 FP16/BF16 約增加四倍。這會增加 GPU 記憶體需求、記憶體傳輸量與計算成本，也可能降低適合低精度運算硬體上的吞吐量。雖然 FP64 的精度較高，但模型不一定能從這麼高的精度得到相應的實際效益。

## 挑戰題

```python
x = 10**1000
```

Python 可以正常運作，是因為 `int` 採用任意精度的整數表示法。`10**1000` 雖然非常大，但它仍然是一個整數，所以 CPython 可以配置足夠的記憶體來保存它。

大型 AI 模型的情況不同。模型通常需要同時保存大量參數，假設有數十億個權重，每個權重多使用幾個 bytes，總記憶體需求就會增加非常多。因此 AI 通常使用固定寬度的 FP32、FP16、BF16，甚至更低精度的格式。

- **記憶體使用量：** FP32 每個值 4 bytes；FP16/BF16 每個值 2 bytes，因此低精度可以大幅減少權重與中間資料的記憶體需求。
- **計算速度：** 現代 GPU 對 FP16/BF16 等格式通常具有高吞吐量的專用硬體，因此大量矩陣乘法可以更快完成。
- **數值精度：** FP64 的精度高於 FP32，而 FP16/BF16 精度較低，所以低精度需要接受捨入誤差。但神經網路通常可以透過適當的訓練方法，在較低精度下維持良好的模型效果。
- **GPU 運算效率：** GPU 適合大量相同格式、固定寬度的資料進行平行運算。Python 任意精度 `int` 的每個數字可能需要不同數量的記憶體，還涉及 Python 物件管理，並不適合作為大型神經網路權重的直接儲存格式。

因此，`10**1000` 與 AI 權重其實是兩種完全不同的需求：前者重視「把一個很大的整數精確保存下來」，後者重視「用合理的精度，以最低的記憶體和最高的硬體吞吐量，同時處理數十億甚至更多的數值」。
